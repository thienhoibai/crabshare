```python
from pathlib import Path

# drive.mount('/content/drive')  # bỏ comment nếu dùng Drive

DATA_YAML = Path('/content/drive/MyDrive/CrabShare-ML/splits/crab/data.yaml')
# DATA_YAML = Path('/content/datasets/crab/data.yaml')

assert DATA_YAML.is_file(), f'Thiếu {DATA_YAML} — chạy prepare_dataset.py rồi chỉnh path'
print('OK', DATA_YAML)
```

```python
# @title Fine-tune
from ultralytics import YOLO

MODEL = 'yolo11n.pt'  # hoặc yolov8n.pt
EPOCHS = 80
IMGSZ = 640
BATCH = 16

model = YOLO(MODEL)
results = model.train(
    data=str(DATA_YAML),
    epochs=EPOCHS,
    imgsz=IMGSZ,
    batch=BATCH,
    project='/content/crabshare-ml',
    name='crab-detector-v1',
    exist_ok=True,
    seed=13,
)
print(results)
```

```python
# @title Xuất ONNX
from ultralytics import YOLO
from pathlib import Path

best = Path('/content/crabshare-ml/crab-detector-v1/weights/best.pt')
model = YOLO(str(best))
export_path = model.export(format='onnx', imgsz=640, simplify=True, opset=12)
print('ONNX:', export_path)
print('Đổi tên thành crab-detector-v1.onnx → GitHub Releases / Drive')
print('Copy vào backend/src/CrabShare.Api/models/ khi chạy API')
```
